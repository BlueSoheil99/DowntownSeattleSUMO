```python
import xml.etree.ElementTree as ET
# from shapely import wkt
import random
```

```python
remaining_percentage = 80
file_name = 'adjusted_alaskanway_od2trips_psrc_vehicle_trips_converted_taz_calibrated_onlyDowntown_fullDay.rou.xml' 
# this file is for a whole 
tree = ET.parse(file_name)
root = tree.getroot()
```

```python
len(list(root[1:]))
```

```python
elements = list(root[1:])  # Convert to a list
num_to_keep = max(1, int(len(elements) * (remaining_percentage / 100)))  # Ensure at least 1 element
selected_elements = set(random.sample(elements, num_to_keep))  # Random selection

```

```python
len(selected_elements)
```

```python
# Remove elements not in selected set
for elem in elements:
    if elem not in selected_elements:
        root.remove(elem)
```

```python
len(root)
```

```python
tree.write(f'reduced_{remaining_percentage}percent_of_'+file_name, encoding='utf-8', xml_declaration=True)
```
